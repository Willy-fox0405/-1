```python
fox=input()
url=("https://zh.wikipedia.org/zh-tw/"+fox)
import requests as rq
response=rq.get(url)
from bs4 import BeautifulSoup as bs
soup=bs(response.content,"html.parser")
box=soup.find("div",class_="mw-parser-output")
boxx=box.find_all("p")
for f in boxx:
  text=f.get_text(strip=True)
  print(text)
```
